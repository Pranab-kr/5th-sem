## Question 1: Genetic Algorithm (Real-Valued, Integer, and Boolean Decision Variables)

```python
import numpy as np
from geneticalgorithm import geneticalgorithm as ga

# (a) Maximize sum of sine values of a real-valued vector X in [0, 10]
def f1(X):
    return -np.sum(np.sin(X))

varbound1 = np.array([[0, 10]] * 3)
model1 = ga(function=f1, dimension=3, variable_type='real', variable_boundaries=varbound1)
model1.run()

# (b) Maximize sum of three integer decision variables in [0, 10]
def f2(X):
    return -np.sum(X)

varbound2 = np.array([[0, 10]] * 3)
model2 = ga(function=f2, dimension=3, variable_type='int', variable_boundaries=varbound2)
model2.run()

# (c) Maximize sum of 30 Boolean decision variables
def f3(X):
    return -np.sum(X)

model3 = ga(function=f3, dimension=30, variable_type='bool')
model3.run()
```

## Question 2: Constrained Genetic Algorithm (Penalty-Based Approach)

```python
import numpy as np
from geneticalgorithm import geneticalgorithm as ga

# (a) Objective function with penalty term for constraint X[0] + X[1] >= 2
def f(X):
    pen = 0
    if X[0] + X[1] < 2:
        pen = 500 + 1000 * (2 - X[0] - X[1])
    return np.sum(X) + pen

# (b) & (c) Configure, execute GA, and display best solution & objective value
varbound = np.array([[0, 10]] * 3)
model = ga(function=f, dimension=3, variable_type='real', variable_boundaries=varbound)
model.run()
```

## Question 4: Hill Climbing Algorithm for f(x) = -x^2 + 5x + 10

```python
"""
Question 4: Hill Climbing Algorithm to Maximize f(x) = -x^2 + 5x + 10.
Matches lab style with per-iteration progress and final best solution output.
"""

import random


# Define the objective function (the function we want to maximize)
def objective_function(x):
    return -(x ** 2) + 5 * x + 10


# Hill climbing algorithm
def hill_climbing(starting_point, step_size=0.1, max_iterations=200):
    current_solution = starting_point
    current_value = objective_function(current_solution)

    for i in range(max_iterations):
        # Generate a neighboring solution
        next_solution = current_solution + random.uniform(-step_size, step_size)
        next_value = objective_function(next_solution)

        # If the new solution is better, move to the new solution
        if next_value > current_value:
            current_solution = next_solution
            current_value = next_value
            print(f"Iteration {i}: Current best solution is {current_solution} with value {current_value}")

    return current_solution, current_value


random.seed(42)
starting_point = random.uniform(-10, 10)
print(f"Initial random starting point: {starting_point} with value {objective_function(starting_point)}")

best_solution, best_value = hill_climbing(starting_point)
print(f"\nBest solution found: {best_solution} with value {best_value}")
```

## Question 5: Gaussian Naive Bayes Classification on Iris Dataset

```python
"""
Question 5: Gaussian Naive Bayes Classification on the Iris Dataset.
Matches lab style using sklearn load_iris, train_test_split, GaussianNB, and accuracy in %.
"""

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn import metrics

# Load the iris dataset
iris = load_iris()

# Store the feature matrix (X) and response vector (y)
X = iris.data
y = iris.target

# Split X and y into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.4, random_state=1)

# Train the model on training set
gnb = GaussianNB()
gnb.fit(X_train, y_train)

# Make predictions on the testing set
y_pred = gnb.predict(X_test)

# Compare actual response values (y_test) with predicted response values (y_pred)
print("Gaussian Naive Bayes model accuracy(in %):", metrics.accuracy_score(y_test, y_pred) * 100)
```

## Question 6: Decision Tree Classification on Diabetes Dataset

```python
"""
Question 6: Decision Tree Classification on the Diabetes Dataset.
Matches lab style using diabetes.csv (Pima Indians Diabetes dataset) with an offline
fallback to sklearn.datasets.load_diabetes() if diabetes.csv is not present.
"""

import os
import numpy as np
import pandas as pd
from sklearn.datasets import load_diabetes
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn import metrics

if os.path.exists("diabetes.csv"):
    col_names = ["pregnant", "glucose", "bp", "skin", "insulin", "bmi", "pedigree", "age", "label"]
    pima = pd.read_csv("diabetes.csv", header=None, names=col_names)
    feature_cols = ["pregnant", "insulin", "bmi", "age", "glucose", "bp", "pedigree"]
    X = pima[feature_cols]
    y = pima.label
else:
    diabetes = load_diabetes()
    X = diabetes.data
    y = (diabetes.target >= np.median(diabetes.target)).astype(int)

# 70% training and 30% test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=1)

# Create and train Decision Tree classifier object
clf = DecisionTreeClassifier(criterion="entropy", max_depth=3)
clf = clf.fit(X_train, y_train)

# Predict the response for test dataset
y_pred = clf.predict(X_test)

print("Accuracy:", metrics.accuracy_score(y_test, y_pred))
```

## Question 7: Exact Inference in Bayesian Belief Network (A -> B -> C)

```python
"""
Question 7: Exact Inference in a Bayesian Belief Network (A -> B -> C).
Defines Conditional Probability Tables (CPTs) for binary nodes A, B, and C,
calculates P(A = True | C = True) using enumeration, and displays the numeric result.
"""

# Bayesian Network structure: A -> B -> C
VARIABLES = ["A", "B", "C"]
PARENTS = {"A": [], "B": ["A"], "C": ["B"]}

# Conditional Probability Tables (CPTs): P(node = True | parent values)
CPTS = {
    "A": {(): 0.3},
    "B": {(True,): 0.8, (False,): 0.2},
    "C": {(True,): 0.9, (False,): 0.1},
}


def prob_given_parents(var, val, evidence):
    parent_vals = tuple(evidence[p] for p in PARENTS[var])
    p_true = CPTS[var][parent_vals]
    return p_true if val else (1.0 - p_true)


def enumerate_all(variables, evidence):
    if not variables:
        return 1.0
    first, rest = variables[0], variables[1:]
    if first in evidence:
        return prob_given_parents(first, evidence[first], evidence) * enumerate_all(rest, evidence)
    total = 0.0
    for val in (True, False):
        extended = dict(evidence)
        extended[first] = val
        total += prob_given_parents(first, val, extended) * enumerate_all(rest, extended)
    return total


def enumeration_ask(query_var, evidence):
    dist = {}
    for val in (True, False):
        extended = dict(evidence)
        extended[query_var] = val
        dist[val] = enumerate_all(VARIABLES, extended)
    norm = sum(dist.values())
    return {val: prob / norm for val, prob in dist.items()}, dist


posterior, unnormalized = enumeration_ask("A", {"C": True})

print("CPTs:")
print("P(A=True) = 0.3, P(A=False) = 0.7")
print("P(B=True | A=True) = 0.8, P(B=True | A=False) = 0.2")
print("P(C=True | B=True) = 0.9, P(C=True | B=False) = 0.1")
print(f"\nP(A=True, C=True)  = {unnormalized[True]:.6f}")
print(f"P(A=False, C=True) = {unnormalized[False]:.6f}")
print(f"P(C=True)          = {unnormalized[True] + unnormalized[False]:.6f}")
print(f"\nP(A = True | C = True) = {posterior[True]:.6f}")
```

## Question 8: Fuzzy Set Operations and Max-Min Composition

```python
"""
Question 8: Basic Operations on Fuzzy Sets and Fuzzy Relations.
Matches lab style for Union, Intersection, Complement, Difference,
Cartesian Product, and Max-Min Composition.
"""

# Define two fuzzy sets
A = [0.2, 0.4, 0.6, 0.8]
B = [0.3, 0.5, 0.7, 0.9]


# 1. Union of two fuzzy sets
def fuzzy_union(A, B):
    return [max(a, b) for a, b in zip(A, B)]


# 2. Intersection of two fuzzy sets
def fuzzy_intersection(A, B):
    return [min(a, b) for a, b in zip(A, B)]


# 3. Complement of a fuzzy set
def fuzzy_complement(A):
    return [round(1 - a, 4) for a in A]


# 4. Difference of two fuzzy sets
def fuzzy_difference(A, B):
    return [round(min(a, 1 - b), 4) for a, b in zip(A, B)]


# 5. Cartesian Product of two fuzzy sets
def cartesian_product(A, B):
    return [[min(a, b) for b in B] for a in A]


# 6. Max-Min Composition of two fuzzy relations
def max_min_composition(R, S):
    rows = len(R)
    cols = len(S[0])
    common = len(S)
    result = []
    for i in range(rows):
        row = []
        for j in range(cols):
            value = max(min(R[i][k], S[k][j]) for k in range(common))
            row.append(value)
        result.append(row)
    return result


# Display results
print("Fuzzy Set A:", A)
print("Fuzzy Set B:", B)

print("\n1. Union:", fuzzy_union(A, B))
print("2. Intersection:", fuzzy_intersection(A, B))
print("3. Complement of A:", fuzzy_complement(A))
print("4. Difference (A - B):", fuzzy_difference(A, B))

print("\n5. Cartesian Product:")
for row in cartesian_product(A, B):
    print(row)

# Fuzzy Relations for Max-Min Composition
R = [
    [0.2, 0.7, 0.5],
    [0.6, 0.4, 0.8],
]
S = [
    [0.5, 0.9],
    [0.7, 0.3],
    [0.6, 0.8],
]

print("\n6. Max-Min Composition (R o S):")
for row in max_min_composition(R, S):
    print(row)
```
